################################################Задание 4: Разделение датасета
# Разделить по уникальным значениям EducationLevel
# Сначала найдем уникальные значения

```python
unique_education = df['EducationLevel'].unique()
print("\nЗадание 4: Уникальные уровни образования:", unique_education)
```

# Создаем словарь DataFrame для каждого уровня

```python
education_dfs = {}
for level in unique_education:
    education_dfs[level] = df[df['EducationLevel'] == level]
```

# Вывод для проверки (пример для одного)

```python
print("\nПример: DataFrame для 'Master ':")
print(education_dfs['Master'].head())
print("\nПример: DataFrame для 'Associate':")
print(education_dfs['Associate'].head())
```

####################################################### Задание 5: Сохранение полученных DataFrame
# сохраняем каждый в отдельный CSV (имя в нижнем регистре, без пробелов)

```python
for level, sub_df in education_dfs.items():
    filename = f"{level.lower().replace(' ', '_')}.csv"
```

# Например: associate.csv

```python
 sub_df.to_csv(filename, index=False)
```
