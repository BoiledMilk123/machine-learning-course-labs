```python
def age_category(age):
    '''
    < 30 -> 1
    >= 30, <55 -> 2
    >= 55 -> 3
    '''
    if age < 30:
        return 1
    elif age < 55:
        return 2
    elif age >= 55:
        return 3

age_categories = [age_category(age) for age in data.Age]
data['Age_category'] = age_categories
```

```python
data['Age_category'] = data['Age'].apply(age_category)
```

```python
#6.Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших людей среди мужчин и женщин?

groupped = data.groupby('Sex')['Survived'].mean() * 100
female_rate = groupped.get('female', 0)
male_rate = groupped.get('male', 0)
print(f"{female_rate:.1f}% среди женщин и {male_rate:.1f}% среди мужчин")
```

```python
# 7. Какое имя наиболее популярно среди пассажиров мужского пола?

def get_first_name(full_name):
    # 1) часть после запятой
    after_comma = full_name.split(',')[1]
     # 2) часть после первой точки ()
    after_dot = after_comma.split('.')[1]
     # 3) первое слово — собственно имя
    first_name = after_dot.strip().split()[0]
    return first_name

male_names = data.loc[data['Sex'] == 'male', 'Name'].apply(get_first_name)

name_counts = male_names.value_counts()

print("Наиболее популярное мужское имя:", name_counts.index[0])
```

```python
# 8. Как средний возраст мужчин / женщин зависит от Pclass? Выберите все правильные утверждения:

result=[]
sort = data.groupby(['Sex', 'Pclass'])['Age'].mean()
if(sort['male',1]>40):
    result.append('В среднем мужчины 1 класса старше 40 лет')
if(sort['female',1]>40):
    result.append('В среднем женщины 1 класса старше 40 лет')
q3=False
for pclass in [1, 2, 3]:
    men_age = sort['male', pclass]
    women_age = sort['female', pclass]
    if (men_age>women_age):q3=True

if q3 ==True: result.append('Мужчины всех классов в среднем старше, чем женщины того же класса')

sort2 = data.groupby('Pclass')['Age'].mean()
if sort2[1]>sort2[2]>sort2[3]:
    result.append('В среднем, пассажиры первого класса старше, чем пассажиры 2-го класса, которые старше, чем пассажиры 3-го класса.')

for q in result:
    print(q)
```
